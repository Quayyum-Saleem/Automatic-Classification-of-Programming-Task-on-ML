```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("D:/Automatic Error Detector/Dataset/dataset_upto500.csv")

# ---------------------------------------
# Class Distribution
# ---------------------------------------

class_counts = df["error_type"].value_counts()

plt.figure(figsize=(10, 6))

class_counts.plot(kind="bar")

plt.title("Distribution of Programming Error Classes")
plt.xlabel("Error Type")
plt.ylabel("Number of Programs")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()


# ---------------------------------------
# Task Distribution
# ---------------------------------------

task_counts = df["task"].value_counts()

plt.figure(figsize=(10, 6))

task_counts.plot(kind="bar")

plt.title("Distribution of Programming Tasks")
plt.xlabel("Programming Task")
plt.ylabel("Number of Programs")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```
