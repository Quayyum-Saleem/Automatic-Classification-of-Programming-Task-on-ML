```python
import ast
import re


# ==========================================
# STEP 1: Preprocess Python Code
# ==========================================

def preprocess_code(code):
    """
    Remove comments and unnecessary blank lines.
    """

    # Remove comments
    code = re.sub(r"#.*", "", code)

    # Split code into lines
    lines = code.splitlines()

    cleaned_lines = []

    # Remove blank lines
    for line in lines:
        if line.strip():
            cleaned_lines.append(line.rstrip())

    # Join the lines again
    cleaned_code = "\n".join(cleaned_lines)

    return cleaned_code


# ==========================================
# STEP 2: Convert Code into AST
# ==========================================

def parse_code(code):
    """
    Convert Python source code into an AST.
    """

    try:
        tree = ast.parse(code)
        return tree

    except SyntaxError:
        return None


# ==========================================
# STEP 3: Find While Loops
# ==========================================

def find_while_loops(tree):

    while_loops = []

    for node in ast.walk(tree):

        if isinstance(node, ast.While):
            while_loops.append(node)

    return while_loops


# ==========================================
# STEP 4: Analyze One While Loop
# ==========================================

def analyze_while_loop(while_node):

    # Get the while-loop condition
    condition = ast.unparse(while_node.test)

    # Get statements inside the loop
    body_statements = []

    for statement in while_node.body:
        body_statements.append(ast.unparse(statement))

    return {
        "condition": condition,
        "body": body_statements
    }


# ==========================================
# STEP 5: Analyze Complete Program
# ==========================================

def analyze_code(code):

    # Preprocess the code
    cleaned_code = preprocess_code(code)

    # Convert code into AST
    tree = parse_code(cleaned_code)

    # If syntax is invalid
    if tree is None:
        return None

    # Find all while loops
    while_loops = find_while_loops(tree)

    # Store information about loops
    loop_information = []

    for loop in while_loops:

        information = analyze_while_loop(loop)

        loop_information.append(information)

    # Create final analysis
    analysis = {
        "cleaned_code": cleaned_code,
        "num_while_loops": len(while_loops),
        "while_loops": loop_information
    }

    return analysis


# ==========================================
# STEP 6: TEST THE PROGRAM
# ==========================================

sample_code = """
# Initialize counter

i = 1

while i <= 10:
    print(i)
    i = i + 1
"""


# Call analyze_code()
result = analyze_code(sample_code)


# ==========================================
# STEP 7: Display Output
# ==========================================

print("=" * 60)
print("AUTOMATIC PYTHON CODE ANALYSIS")
print("=" * 60)

print("\nOriginal Code:")
print(sample_code)

print("\nCleaned Code:")
print(result["cleaned_code"])

print("\nNumber of While Loops:")
print(result["num_while_loops"])

print("\nWhile Loop Information:")

for i, loop in enumerate(result["while_loops"], start=1):

    print("\nWhile Loop", i)

    print("Condition:")
    print(loop["condition"])

    print("Body:")

    for statement in loop["body"]:
        print("   ", statement)

print("\n" + "=" * 60)
print("ANALYSIS COMPLETED")
print("=" * 60)
```

```python
import pandas as pd
import ast
import re


# ============================================================
# MODULE 2: PYTHON CODE PREPROCESSING & AST ANALYSIS
# ============================================================


# ============================================================
# STEP 1: LOAD DATASET
# ============================================================

DATASET_PATH = "../dataset/validated_programs.csv"
OUTPUT_PATH = "../dataset/processed_programs.csv"

df = pd.read_csv(DATASET_PATH)

print("=" * 70)
print("MODULE 2 - PYTHON CODE PREPROCESSING & AST ANALYSIS")
print("=" * 70)

print("\nDataset loaded successfully!")
print("Number of programs:", len(df))


# ============================================================
# STEP 2: PREPROCESS PYTHON CODE
# ============================================================

def preprocess_code(code):
    """
    Remove comments and unnecessary blank lines.
    """

    # Convert code to string
    code = str(code)

    # Remove comments
    code = re.sub(r"#.*", "", code)

    # Split code into individual lines
    lines = code.splitlines()

    cleaned_lines = []

    # Remove blank lines
    for line in lines:

        if line.strip():

            # Remove unnecessary spaces at the end
            cleaned_lines.append(line.rstrip())

    # Join lines back together
    cleaned_code = "\n".join(cleaned_lines)

    return cleaned_code


# ============================================================
# STEP 3: PARSE CODE USING AST
# ============================================================

def parse_code(code):
    """
    Convert Python source code into an Abstract Syntax Tree.
    """

    try:

        tree = ast.parse(code)

        return tree

    except SyntaxError:

        return None


# ============================================================
# STEP 4: FIND WHILE LOOPS
# ============================================================

def find_while_loops(tree):

    while_loops = []

    for node in ast.walk(tree):

        if isinstance(node, ast.While):

            while_loops.append(node)

    return while_loops


# ============================================================
# STEP 5: FIND VARIABLES
# ============================================================

def find_variables(tree):

    variables = set()

    for node in ast.walk(tree):

        # Variables used in assignments
        if isinstance(node, ast.Name):

            variables.add(node.id)

    return list(variables)


# ============================================================
# STEP 6: FIND ASSIGNMENTS
# ============================================================

def find_assignments(tree):

    assignments = []

    for node in ast.walk(tree):

        if isinstance(node, ast.Assign):

            try:

                assignments.append(ast.unparse(node))

            except:

                pass

    return assignments


# ============================================================
# STEP 7: FIND INITIALIZATION STATEMENTS
# ============================================================

def find_initializations(tree):

    initializations = []

    for node in ast.walk(tree):

        if isinstance(node, ast.Assign):

            try:

                statement = ast.unparse(node)

                initializations.append(statement)

            except:

                pass

    return initializations


# ============================================================
# STEP 8: ANALYZE WHILE LOOP
# ============================================================

def analyze_while_loop(while_node):

    # -----------------------------------------
    # Get loop condition
    # -----------------------------------------

    try:

        condition = ast.unparse(while_node.test)

    except:

        condition = ""


    # -----------------------------------------
    # Get loop body
    # -----------------------------------------

    body_statements = []

    for statement in while_node.body:

        try:

            body_statements.append(
                ast.unparse(statement)
            )

        except:

            pass


    # -----------------------------------------
    # Find variables used in condition
    # -----------------------------------------

    condition_variables = []

    for node in ast.walk(while_node.test):

        if isinstance(node, ast.Name):

            condition_variables.append(node.id)


    # -----------------------------------------
    # Find update statements
    # -----------------------------------------

    update_statements = []

    for statement in while_node.body:

        # Example:
        # i = i + 1

        if isinstance(statement, ast.Assign):

            try:

                text = ast.unparse(statement)

                update_statements.append(text)

            except:

                pass


        # Example:
        # i += 1

        elif isinstance(statement, ast.AugAssign):

            try:

                text = ast.unparse(statement)

                update_statements.append(text)

            except:

                pass


    # -----------------------------------------
    # Detect condition operator
    # -----------------------------------------

    condition_operator = ""

    if isinstance(while_node.test, ast.Compare):

        if len(while_node.test.ops) > 0:

            operator = while_node.test.ops[0]

            if isinstance(operator, ast.Lt):
                condition_operator = "<"

            elif isinstance(operator, ast.LtE):
                condition_operator = "<="

            elif isinstance(operator, ast.Gt):
                condition_operator = ">"

            elif isinstance(operator, ast.GtE):
                condition_operator = ">="

            elif isinstance(operator, ast.Eq):
                condition_operator = "=="

            elif isinstance(operator, ast.NotEq):
                condition_operator = "!="


    return {

        "condition": condition,

        "condition_variables":
            condition_variables,

        "condition_operator":
            condition_operator,

        "body":
            body_statements,

        "update_statements":
            update_statements,

        "body_size":
            len(body_statements)

    }


# ============================================================
# STEP 9: ANALYZE COMPLETE PROGRAM
# ============================================================

def analyze_code(code):

    # -----------------------------------------
    # Preprocess code
    # -----------------------------------------

    cleaned_code = preprocess_code(code)


    # -----------------------------------------
    # Convert code into AST
    # -----------------------------------------

    tree = parse_code(cleaned_code)


    # -----------------------------------------
    # Handle invalid syntax
    # -----------------------------------------

    if tree is None:

        return {

            "cleaned_code": cleaned_code,

            "syntax_valid": 0,

            "num_variables": 0,

            "num_assignments": 0,

            "num_initializations": 0,

            "num_while_loops": 0,

            "while_conditions": "",

            "condition_variables": "",

            "condition_operators": "",

            "update_statements": "",

            "loop_body_size": 0,

            "loop_control_variables": ""

        }


    # -----------------------------------------
    # Find variables
    # -----------------------------------------

    variables = find_variables(tree)


    # -----------------------------------------
    # Find assignments
    # -----------------------------------------

    assignments = find_assignments(tree)


    # -----------------------------------------
    # Find initializations
    # -----------------------------------------

    initializations = find_initializations(tree)


    # -----------------------------------------
    # Find while loops
    # -----------------------------------------

    while_loops = find_while_loops(tree)


    # -----------------------------------------
    # Store loop information
    # -----------------------------------------

    loop_information = []


    for loop in while_loops:

        information = analyze_while_loop(loop)

        loop_information.append(information)


    # -----------------------------------------
    # Extract loop conditions
    # -----------------------------------------

    conditions = []

    for information in loop_information:

        conditions.append(
            information["condition"]
        )


    # -----------------------------------------
    # Extract condition variables
    # -----------------------------------------

    condition_variables = []

    for information in loop_information:

        condition_variables.extend(
            information["condition_variables"]
        )


    # Remove duplicates
    condition_variables = list(
        set(condition_variables)
    )


    # -----------------------------------------
    # Extract condition operators
    # -----------------------------------------

    condition_operators = []

    for information in loop_information:

        operator = information[
            "condition_operator"
        ]

        if operator:

            condition_operators.append(operator)


    # -----------------------------------------
    # Extract update statements
    # -----------------------------------------

    update_statements = []

    for information in loop_information:

        update_statements.extend(
            information["update_statements"]
        )


    # -----------------------------------------
    # Calculate total loop body size
    # -----------------------------------------

    total_body_size = 0

    for information in loop_information:

        total_body_size += information[
            "body_size"
        ]


    # -----------------------------------------
    # Loop control variables
    # -----------------------------------------

    loop_control_variables = condition_variables.copy()


    # -----------------------------------------
    # Return complete analysis
    # -----------------------------------------

    analysis = {

        "cleaned_code":
            cleaned_code,

        "syntax_valid":
            1,

        "num_variables":
            len(variables),

        "num_assignments":
            len(assignments),

        "num_initializations":
            len(initializations),

        "num_while_loops":
            len(while_loops),

        "while_conditions":
            " | ".join(conditions),

        "condition_variables":
            " | ".join(condition_variables),

        "condition_operators":
            " | ".join(condition_operators),

        "update_statements":
            " | ".join(update_statements),

        "loop_body_size":
            total_body_size,

        "loop_control_variables":
            " | ".join(loop_control_variables)

    }


    return analysis


# ============================================================
# STEP 10: PROCESS ALL PROGRAMS
# ============================================================

print("\nProcessing all programs...")

processed_data = []


for index, row in df.iterrows():

    code = row["code"]

    result = analyze_code(code)


    # Combine original dataset information
    processed_row = {

        "id":
            row["id"],

        "task_id":
            row["task_id"],

        "task":
            row["task"],

        "error_type":
            row["error_type"],

        "code":
            code,

        "cleaned_code":
            result["cleaned_code"],

        "syntax_valid":
            result["syntax_valid"],

        "num_variables":
            result["num_variables"],

        "num_assignments":
            result["num_assignments"],

        "num_initializations":
            result["num_initializations"],

        "num_while_loops":
            result["num_while_loops"],

        "while_conditions":
            result["while_conditions"],

        "condition_variables":
            result["condition_variables"],

        "condition_operators":
            result["condition_operators"],

        "update_statements":
            result["update_statements"],

        "loop_body_size":
            result["loop_body_size"],

        "loop_control_variables":
            result["loop_control_variables"]

    }


    processed_data.append(processed_row)


    # Display progress
    if (index + 1) % 50 == 0:

        print(
            "Processed:",
            index + 1,
            "programs"
        )


# ============================================================
# STEP 11: CREATE PROCESSED DATAFRAME
# ============================================================

processed_df = pd.DataFrame(
    processed_data
)


# ============================================================
# STEP 12: DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 70)

print("PROCESSING COMPLETED")

print("=" * 70)

print(
    "\nTotal programs processed:",
    len(processed_df)
)

print(
    "Valid programs:",
    processed_df["syntax_valid"].sum()
)

print(
    "Invalid programs:",
    len(processed_df)
    - processed_df["syntax_valid"].sum()
)


# ============================================================
# STEP 13: DISPLAY SAMPLE
# ============================================================

print("\nSample Processed Program:")

sample = processed_df.iloc[0]

print("\nProgram ID:")
print(sample["id"])

print("\nTask:")
print(sample["task"])

print("\nError Type:")
print(sample["error_type"])

print("\nOriginal Code:")
print(sample["code"])

print("\nCleaned Code:")
print(sample["cleaned_code"])

print("\nNumber of Variables:")
print(sample["num_variables"])

print("\nNumber of Assignments:")
print(sample["num_assignments"])

print("\nNumber of While Loops:")
print(sample["num_while_loops"])

print("\nWhile Condition:")
print(sample["while_conditions"])

print("\nCondition Variables:")
print(sample["condition_variables"])

print("\nCondition Operator:")
print(sample["condition_operators"])

print("\nUpdate Statements:")
print(sample["update_statements"])

print("\nLoop Body Size:")
print(sample["loop_body_size"])

print("\nLoop Control Variables:")
print(sample["loop_control_variables"])


# ============================================================
# STEP 14: SAVE PROCESSED DATASET
# ============================================================

processed_df.to_csv(
    OUTPUT_PATH,
    index=False
)


print("\n" + "=" * 70)

print("PROCESSED DATASET SAVED SUCCESSFULLY!")

print("=" * 70)

print("\nFile:")
print(OUTPUT_PATH)
```
