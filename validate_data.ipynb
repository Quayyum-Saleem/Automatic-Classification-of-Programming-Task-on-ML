```python
import pandas as pd
import ast

# ---------------------------------------
# 1. Load Dataset
# ---------------------------------------

file_path = "D:/Automatic Error Detector/Dataset/dataset_upto500.csv"

df = pd.read_csv("D:/Automatic Error Detector/Dataset/dataset_upto500.csv")

print("=" * 50)
print("DATASET VALIDATION")
print("=" * 50)
print(df.head())


```

```python
# ---------------------------------------
# 2. Basic Information
# ---------------------------------------

print("\n1. Dataset Shape")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())
```

```python
# ---------------------------------------
# 3. Missing Values
# ---------------------------------------

print("\n2. Missing Values")
print(df.isnull().sum())
```

```python
# ---------------------------------------
# 4. Duplicate Programs
# ---------------------------------------

print("\n3. Duplicate Programs")

duplicates = df["code"].duplicated().sum()

print("Duplicate programs:", duplicates)
```

```python
# ---------------------------------------
# 5. Class Distribution
# ---------------------------------------

print("\n4. Error Class Distribution")

print(df["error_type"].value_counts())
```

```python
# ---------------------------------------
# 6. Task Distribution
# ---------------------------------------

print("\n5. Task Distribution")

print(df["task"].value_counts())
```

```python
# ---------------------------------------
# 7. Syntax Validation
# ---------------------------------------

print("\n6. Python Syntax Validation")

valid_count = 0
invalid_count = 0

invalid_rows = []

for index, code in enumerate(df["code"]):

    try:
        ast.parse(code)
        valid_count += 1

    except SyntaxError:
        invalid_count += 1
        invalid_rows.append(index)

print("Valid programs:", valid_count)
print("Invalid programs:", invalid_count)
```

```python
# ---------------------------------------
# 8. Display Samples
# ---------------------------------------

print("\n7. Sample Programs")

for i in range(5):

    print("\nProgram ID:", df.iloc[i]["id"])
    print("Task:", df.iloc[i]["task"])
    print("Error Type:", df.iloc[i]["error_type"])

    print("Code:")
    print(df.iloc[i]["code"])
```

```python
# ---------------------------------------
# 9. Save Validated Dataset
# ---------------------------------------

if invalid_count == 0:

    output_file = "../dataset/validated_programs.csv"

    df.to_csv(output_file, index=False)

    print("\nValidated dataset saved successfully!")
    print(output_file)

else:

    print("\nWARNING:")
    print("Some programs contain syntax errors.")

print("\n" + "=" * 50)
print("VALIDATION COMPLETED")
print("=" * 50)
```
